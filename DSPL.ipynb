### Milestone 2: Core Python Skills & OOP
* **Goal:** Model sales data using Object-Oriented Programming and handle errors gracefully.
* **Key Steps:**
  1. **Custom Exception:** Defined `InvalidTransactionError(Exception)` to validate unit price and quantity.
  2. **OOP Hierarchy:** Created `SaleTransaction` base class, with `MemberSale` (5% discount via `super()`) and `NormalSale` subclasses.
  3. **Functional Python:** Used List Comprehensions, `map()`, `filter()`, and dictionary sorting (`lambda key`) to process transaction samples.

* **AI Usege**: Refactored MemberSale discounts via super() and built custom exception handling (InvalidTransactionError)

```python
print ("___Milestone 2: Python coor skills , OOP and Excepation ___")
sub_df = df.head(20).copy()
#List comrahntion Example
high_ratings = [rating for rating in sub_df['Rating'] if rating > 0.8]
sub_df['Branch_lower'] = list(map(lambda x:x.lower() , sub_df['Branch']))
#Filter transiction with quantity greater than  5
large_quantity = list(filter(lambda x :x>5 , sub_df['Quantity']))
def sort_dicT_by_values(d):
    return dict(sorted(d.items(),key = lambda item:item[1], reverse = True))
customer_counts = sub_df['Customer type'].value_counts().to_dict()
sorted_custumer = sort_dicT_by_values(customer_counts)

class InvalidTransactionError(Exception):
    pass
#Base class (pearent)
class SaleTransction:
    def __init__(self, invoice_id, product_line, unit_price, quantity, rating):
        if unit_price <= 0 or quantity<=0 :
            raise InvalidTransactionError("Unite price and quantity must be greater than 0")
        if not product_line or pd.isna(product_line):
            raise  InvalidTransactionError("Product line can not be missing")
        if not(0<=rating<=10):
            raise InvalidTransactionError("Ratig must be between 0 annd 10")
        
        self.invoice_id =  invoice_id
        self.product_line = product_line
        self._unite_price = unit_price
        self._quantity = quantity
        self.rating = rating

    def calculate_total(self):
        return self._unite_price * self._quantity
#Child class for member customers        
class MemberSale(SaleTransction):
    def calculate_total(self):
        base_total = super().calculate_total()
        return base_total * 0.95
#Child class for normal customers    
class NormalSale (SaleTransction):
    def calculate_total(self):
        base_total = super().calculate_total()
        return base_total
object_list = []     
rows_to_creat = sub_df.iloc[10:21]
for index, row in rows_to_creat.iterrows():
    try:
        if row['Customer type'] == 'Member':
            obj = MemberSale(row['Invoice ID'], row['Product line'], row['Unit price'], row['Quantity'], row['Rating'])
        else:
            obj = NormalSale(row['Invoice ID'], row['Product line'], row['Unit price'], row['Quantity'], row['Rating'])
        object_list.append(obj)
    except InvalidTransactionError as e:
        print(f"Skipped row {index}: {e}")
print(f"Successfully created {len(object_list)}transaction objects.")
```

### Milestone 4: Advanced Pandas Analysis & Excel Export
* **Goal:** Aggregate results for management and export the final report.
* **Key Steps:**
  1. Built a **Pivot Table** comparing `Sales` and `Rating` across `Product line` and `Customer type`.
  2. Grouped data by branch/city using `groupby()` to evaluate performance.
  3. Exported the cleaned summary to an Excel file using `to_excel()`.
* **AI Usege**:How to handle with Data farame , orgnizing data in EXCEL and automated clean Excel export.

```python
print ("___Milestone 4: Pandas Anlysis and Exel Exports ___")
#Clean the dataset
df.columns = df.columns.str.strip()
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
df['Date'] = pd.to_datetime(df['Date'])

highest_branch = df.groupby('Branch')['Sales'].sum().idxmax()
highest_product_line = df.groupby('Product line')['Rating'].mean().idxmax()
popular_payment = df['Payment'].mode()[0]
#Answer anlysis quastions using panda
summary_answers = pd.DataFrame({
    'Question': [
        'Which branch has the highest total sales?',
        'Which product line has the highest average rating?',
        'Which payment method is used most often?'
    ],
    'Answer': [highest_branch, highest_product_line, popular_payment]
})
#Create a pivot table
pivot_rep = df.pivot_table(
    index='Product line',
    columns='Customer type',
    values=['Sales', 'Rating'],
    aggfunc='mean'
)
#Export cleaned data and anlysis to an Excel files
with pd.ExcelWriter('Supermarket_Sales_Report.xlsx-')as writer :
    df.to_excel(writer, sheet_name='Cleaned_Data',index=False)
    pivot_rep.to_excel(writer, sheet_name='Pivot_Analysis')
    summary_answers.to_excel(writer, sheet_name='Summary_Answers', index=False)
    print("Excel report 'Supermarket_Sales_Report.xlsx' has been successfully created")
```
